## Data and network requirements
The prediction notebooks default to the sample GeoJSON in `examples/data/` (Port Phillip Bay, with Madrid and Austin, Texas area alternatives, and Cabauw and Railroad Valley point samples). Launch Jupyter from the repository root or `notebooks/`. Set `EO_OVERPASS_WORKSPACE` to choose a different workspace for local caches and reports; its `DATA/` and `OUTPUT/` directories are created as needed. Edit the ROI input to use your own vector file, point or bounding box. TLE and ESA plan downloads need network access. For offline work supply saved elements to `fetch_tles(..., tle_text=..., offline=True)` and load local plans. See the README for the programmatic and CLI workflows.
